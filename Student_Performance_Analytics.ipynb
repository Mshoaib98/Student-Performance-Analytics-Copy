```python
import zipfile
import os
from google.colab import files

# 1. Prompt to upload the zip file
print("Please select and upload your ZIP file:")
uploaded_zip = files.upload()

# 2. Extract the zip file
if uploaded_zip:
    zip_filename = list(uploaded_zip.keys())[0]
    extract_dir = "extracted_py_files"

    with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print(f"\nSuccessfully extracted {zip_filename} to '{extract_dir}/'\n")

    # 3. Find and print contents of the .py files
    py_files = []
    for root, dirs, files_list in os.walk(extract_dir):
        for file in files_list:
            if file.endswith('.py'):
                py_files.append(os.path.join(root, file))

    print(f"Found {len(py_files)} Python file(s): {[os.path.basename(f) for f in py_files]}\n")

    for py_file in py_files:
        print("=" * 40)
        print(f" File: {os.path.basename(py_file)}")
        print("=" * 40)
        with open(py_file, 'r', encoding='utf-8', errors='ignore') as f:
            print(f.read())
        print("\n")
else:
    print("No file was uploaded.")
```
